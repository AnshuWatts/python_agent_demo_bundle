```python
import os
from dotenv import load_dotenv
from google import genai

def main() -> None:
    # Load variables from the .env file
    load_dotenv()

     # Get the Gemini API key from the environment
    api_key = os.getenv("GOOGLE_API_KEY")

    # Make sure the API key exists
    if not api_key:
        raise RuntimeError(
            "GOOGLE_API_KEY was not found. Create a .env file and add your key first."
        )

    # Create Gemini client
    client = genai.Client(api_key=api_key)

     # Send a request to Gemini
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents="Reply with exactly these two words: environment ready",
    )

    # Display Gemini's response
    print("Model reply:")
    print(response.text)

    # Get token usage information
    usage = getattr(response, "usage_metadata", None)
    if usage:
        prompt_tokens = getattr(usage, "prompt_token_count", "n/a")
        response_tokens = getattr(usage, "candidates_token_count", "n/a")
        print(f"Prompt tokens: {prompt_tokens}")
        print(f"Response tokens: {response_tokens}")

if __name__ == "__main__":
    main()
```

1. Import os
import os

os is Python's built-in module for interacting with the operating system.

Here we use it to read the API key stored in the environment:

os.getenv("GOOGLE_API_KEY")


2. Import load_dotenv

from dotenv import load_dotenv

This comes from the python-dotenv package you installed.

It allows Python to read variables from your .env file.

For example, your .env file contains:

GOOGLE_API_KEY=your_actual_api_key

Then:

load_dotenv()

loads that value into the environment.


3. Import Google's GenAI SDK

from google import genai

This comes from the:

google-genai

package.

It allows your Python program to communicate with Google's Gemini API.

The basic architecture is:

Python

   ↓
google-genai

   ↓
Gemini API

   ↓
Gemini model


4. Create the main() function

def main() -> None:

This defines the main function of your program.

-> None means the function isn't expected to return a value.

Everything related to setting up and testing Gemini is placed inside this function.

load_dotenv()

6. Retrieve the API key

api_key = os.getenv("GOOGLE_API_KEY")

This is an important distinction.

You do not put your actual API key inside getenv().

Correct:

os.getenv("GOOGLE_API_KEY")

Incorrect:

os.getenv("your_actual_api_key")

The flow is:

.env

 │ GOOGLE_API_KEY=ABC123...

7. Check whether the API key exists

if not api_key:
    raise RuntimeError(
        "GOOGLE_API_KEY was not found. "
        "Create a .env file and add your key first."
    )

If Python cannot find:

GOOGLE_API_KEY

then api_key will be None.

The program stops and gives you a meaningful error instead of continuing and producing a less useful API error.


8. Create the Gemini client

client = genai.Client(api_key=api_key)

This creates a Gemini API client.

You can think of client as the object your Python program uses to communicate with Gemini.

Python program

      │

      ▼

    client

      │

      ▼

 Gemini API


9. Send a prompt to Gemini

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Reply with exactly these two words: environment ready",
)

This is the actual Gemini API call.

There are two important arguments.

Model

model="gemini-3.6-flash"

This tells Gemini which model should process the request.

Prompt

contents="Reply with exactly these two words: environment ready"

This is the input given to the model.

You're deliberately asking for a very simple response so you can verify that the API connection works.

Expected response:

environment ready


10. Print the model response

print("Model reply:")

print(response.text)

response contains the API response.

The generated text is available through:

response.text

So you should see:

Model reply:

environment ready


11. Get token usage

This part is slightly more advanced:

usage = getattr(response, "usage_metadata", None)

The Gemini response can contain information about how many tokens were used.

For example:

Prompt tokens: 9

Response tokens: 3

getattr() safely attempts to retrieve:

response.usage_metadata

If it doesn't exist, it returns:

None

instead of crashing.


12. Check whether usage information exists

if usage:

This means:

If Gemini returned usage information, process it.


13. Get input/prompt tokens

prompt_tokens = getattr(
    usage,
    "prompt_token_count",
    "n/a",
)

This retrieves the number of tokens used by your input.

For example:

Prompt tokens: 10


14. Get output/response tokens

response_tokens = getattr(
    usage,
    "candidates_token_count",
    "n/a",
)

This retrieves the number of tokens generated by the model.

For example:

Response tokens: 3


15. Print token usage

print(f"Prompt tokens: {prompt_tokens}")

print(f"Response tokens: {response_tokens}")

The f before the string allows Python to insert variable values.

For example:

prompt_tokens = 10

results in:

Prompt tokens: 10
